```python
from pandas import *
from matplotlib.pyplot import *
from numpy import *
```

```python
data=read_csv(r'C:\Users\ayush\Downloads\Python Projects\python Datasets\customers.csv')
data
```

```python
data.info()
```

```python
x=data['Age'].values
y=data['Spending Score (1-100)'].values
F=np.array(list(zip(x,y)))
G=scatter(x,y,c='black')
```

```python
from sklearn.cluster import KMeans
```

```python
ssd=[]
for i in range(1,11):
    kmeans=KMeans(n_clusters=i)
    kmeans.fit(F)
    ssd.append(kmeans.inertia_)
```

```python
plot(ssd)
```

```python
kmeans=KMeans(n_clusters=3)
kmeans.fit(F)
```

```python
kmeans.labels_
```

```python
data['CID']=kmeans.labels_
```

```python
data
```

```python
centroids=kmeans.cluster_centers_
```
