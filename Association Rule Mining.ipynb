```python
from pandas import *
from numpy import *
from mlxtend.frequent_patterns import apriori, association_rules  
from apyori import apriori
```

```python
import chardet
with open('Retail.csv', 'rb') as rawdata:
    result = chardet.detect(rawdata.read(100000))
result
```

```python
data=read_csv('Retail.csv',encoding='ISO-8859-9',header=None)
```

```python
data.head()
```

```python
data.shape
```

```python
records=[]
for i in range(0,19):
    records.append([str(data.values[i,j]) for j in range(0,6)])
```

```python
records[2]
```

```python
association_rules=apriori(records,min_support=0.0045, min_confidence=0.2,min_lift=3,min_length=1)
association_results=list(association_rules)
```
